```python
import os
import librosa

folder = "data/clips_noisy"

for root, _, files in os.walk(folder):
    for file in files:
        if file.endswith(".wav"):
            path = os.path.join(root, file)

            y, sr = librosa.load(path, sr=22050)
            if len(y)/sr != 5.0:
                print(os.path.dirname(path), file, len(y), len(y)/sr)
            
```

```python
import os
import numpy as np

widths = []

for root, _, files in os.walk("data/spectrograms"):
    for file in files:
        if file.endswith(".npy"):
            spec = np.load(os.path.join(root, file))
            widths.append(spec.shape)

print("Unique shapes:")
print(sorted(set(widths)))
```

```python
import os
import numpy as np
from collections import Counter

shapes = Counter()

for root, _, files in os.walk("data/spectrograms"):
    for file in files:
        if file.endswith(".npy"):
            spec = np.load(os.path.join(root, file))
            shapes[spec.shape] += 1

print(shapes)
```

```python
import os
import numpy as np

for root, _, files in os.walk("data/spectrograms"):
    for file in files:
        if file.endswith(".npy"):
            path = os.path.join(root, file)
            spec = np.load(path)

            if spec.shape != (128, 216):
                print(spec.shape, path)
```
